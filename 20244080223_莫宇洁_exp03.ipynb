实验1：正则基础匹配

```python
import re
text = "cat on mat; later: cats and dogs"
print(re.findall(r"\d+", text))      # 数字
print(re.findall(r"\s+", text))      # 空白
print(re.findall(r"\bcat\b", text))  # 独立单词 cat

s = "cat dog bird CAT"
print(re.findall(r"cat|dog", s))   # 注意默认区分大小写，CAT 不匹配
print(re.findall(r"a|aa|aaa", "aaaa"))
print(re.findall(r"aaa|aa|a", "aaaa"))

s = "ct cat caat caaat"
for p in [r"ca*t", r"ca+t", r"ca?t", r"ca{2}t", r"ca{1,2}t"]:
    print(p, re.findall(p, s))
```

实验2：分组、回溯引用与替换

```python
s = "联系: alice@example.com; 备选: bob@test.org"
print(re.findall(r"(\w+)@(\w+\.\w+)", s))
```

```python
s = "cat cat, dog cat, bird bird"
print(re.findall(r"\b(\w+)\s+\1\b", s))
```

```python
s = "2024-05-01\n2024/05/02\n2024.05.03"
pat = r"(\d{4})[-/.](\d{2})[-/.](\d{2})"
new = re.sub(pat, r"\2-\3-\1", s)
print("转换前:\n", s)
print("转换后:\n", new)
```

```python
s = "联系我: 123-456-7890 或 987.654.3210"
pat = r"(\d{3})[-.](\d{3})[-.](\d{4})"
new = re.sub(pat, r"(\1) \2-\3", s)
print(new)
```

实验3：BeautifulSoup 解析 HTML

```python
from bs4 import BeautifulSoup
html_doc = """<html><head><title>实验课程</title></head><body>
<h1>数据分析</h1>
<p class="course">第六讲: 正则表达式</p>
<p class="course">第七讲: 结构化数据</p>
<a href="https://example.org/lecture06">讲义六</a>
<a href="https://example.org/lecture07">讲义七</a>
</body></html>"""
soup = BeautifulSoup(html_doc, "html.parser")
print("标题:", soup.title.get_text())
h1 = soup.h1
print("h1 标签名:", h1.name, "文本:", h1.get_text())
```

```python
for p in soup.find_all("p", class_="course"):
    print("段落:", p.get_text())
for a in soup.find_all("a"):
    print("链接文字:", a.get_text(), "href:", a["href"])
```

```python
print("整个文档可见文本:", repr(soup.get_text()))
print("单个段落:", repr(soup.find("p", class_="course").get_text()))
print("单个链接:", repr(soup.find("a").get_text()))
```
