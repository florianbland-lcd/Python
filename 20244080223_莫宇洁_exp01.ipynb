```python
# ========= 实验2：函数、条件、循环与递归 =========

# 1. score_level 函数
def score_level(score):
    if not (0 <= score <= 100):
        raise ValueError("分数必须在 0~100 之间")
    if score >= 90:
        return "优秀"
    elif score >= 80:
        return "良好"
    elif score >= 60:
        return "及格"
    else:
        return "不及格"

print("--- 测试 score_level ---")
print(f"95分: {score_level(95)}")
print(f"85分: {score_level(85)}")
print(f"75分: {score_level(75)}")
print(f"55分: {score_level(55)}")
try:
    score_level(105)
except ValueError as e:
    print(f"捕获异常: {e}")


# 2. 分别使用for和while计算1到100的整数和
print("\n--- 计算 1 到 100 的和 ---")
sum_for = sum(range(1, 101))
print(f"for循环结果: {sum_for}")

sum_while = 0
k = 1
while k <= 100:
    sum_while += k
    k += 1
print(f"while循环结果: {sum_while}")


# 3. 递归版和迭代版阶乘函数
def fact_recursive(n):
    if n == 0:
        return 1
    return n * fact_recursive(n - 1)

def fact_iterative(n):
    result = 1
    for i in range(2, n + 1):
        result *= i
    return result

print("\n--- 测试阶乘 (n=0, 1, 5, 10) ---")
for n in [0, 1, 5, 10]:
    print(f"n={n}: 递归={fact_recursive(n)}, 迭代={fact_iterative(n)}")


# 4. summary 函数
def summary(values):
    return min(values), max(values), sum(values) / len(values)

print("\n--- 测试 summary ---")
lo, hi, avg = summary([85, 58, 92, 76, 60, 89])
print(f"最小值: {lo}, 最大值: {hi}, 平均值: {avg:.2f}")


# 5. sum_all 函数
def sum_all(*args):
    return sum(args)

print("\n--- 测试 sum_all ---")
print(f"无参数: {sum_all()}")
print(f"单参数: {sum_all(5)}")
print(f"多参数: {sum_all(1, 2, 3, 4)}")
```

```python
# ========= 实验3：列表、字典、元组与基础统计 =========

names = ["Alice", "Bob", "Carol", "David", "Eva", "Frank"]
scores = [85, 58, 92, 76, 60, 89]

# 1. 构建字典并输出
print("--- 1. 姓名到成绩的字典 ---")
score_dict = dict(zip(names, scores))
for name, score in score_dict.items():
    print(f"{name}: {score}")

# 2. 使用 enumerate 输出（序号从1开始）
print("\n--- 2. enumerate 输出 ---")
for idx, (name, score) in enumerate(score_dict.items(), start=1):
    print(f"序号 {idx}: {name} - {score}分")

# 3. 列表推导式
print("\n--- 3. 列表推导式 ---")
passed_names = [name for name, score in zip(names, scores) if score >= 60]
squared_scores = [score ** 2 for score in scores]
print(f"及格学生: {passed_names}")
print(f"成绩平方: {squared_scores}")

# 4. 基础统计与格式化
print("\n--- 4. 基础统计 ---")
count = len(scores)
avg_score = sum(scores) / count
max_score = max(scores)
min_score = min(scores)
pass_rate = len(passed_names) / count
print(f"总人数: {count}")
print(f"平均分: {avg_score:.2f}")    # 保留两位小数
print(f"最高分: {max_score}")
print(f"最低分: {min_score}")
print(f"及格率: {pass_rate:.0%}")    # 显示为百分数

# 5. 比较 alias 与 copied (使用新的演示列表，不改变原有成绩)
print("\n--- 5. 赋值与浅拷贝的区别 ---")
demo_scores = [70, 80, 90]
alias = demo_scores
copied = demo_scores.copy()

alias[0] = 100
print(f"原列表: {demo_scores}")  # 会跟着 alias 变，因为指向同一对象
print(f"copied 副本: {copied}")     # 不受影响，因为是独立的拷贝

# 6. 元组不可变性与 zip 长度不同的结果
print("\n--- 6. 元组与 zip 特性 ---")
student_record = ("Alice", 85)
try:
    student_record[1] = 90
except TypeError:
    print("捕获异常: 元组元素不能直接重新赋值")

zip_diff_len = list(zip([1, 2, 3], ["A", "B"]))
print(f"长度不同序列的 zip 结果: {zip_diff_len}")  # 以短的为准
```

```python
# ========= 实验4：文件读写、异常处理与模块 =========
import os
import pickle

# 1. 查看工作目录并创建数据文件夹
print("--- 1. 目录操作 ---")
print("当前工作目录:", os.getcwd())
data_dir = os.path.join(os.getcwd(), "exp01_data")
os.makedirs(data_dir, exist_ok=True)
print("已创建/确认数据目录:", data_dir)

# 2. 写入文本文件
file_path = os.path.join(data_dir, "students.txt")
content = """name,score
Alice,85
Bob,58
Carol,92
David,76
Eva,60
Frank,89
Grace,abc
Henry,105
Incomplete
"""
with open(file_path, "w", encoding="utf-8") as f:
    f.write(content)
print("\n--- 2. 文件写入完毕 ---")

# 3 & 4. 定义解析函数并按行处理文件，捕获异常
def parse_student(line):
    parts = line.strip().split(",")
    if len(parts) != 2:
        raise ValueError("字段数量不为 2")
    name, score_s = parts[0].strip(), parts[1].strip()
    if not name:
        raise ValueError("姓名为空")
    score = int(score_s) # 这里可能会报错
    if not (0 <= score <= 100):
        raise ValueError("成绩超出 0~100")
    return {"name": name, "score": score}

valid_records = []
invalid_count = 0
print("\n--- 3 & 4. 解析文件数据 ---")
with open(file_path, encoding="utf-8") as f:
    next(f) # 跳过头
    for lineno, line in enumerate(f, start=2):
        if not line.strip(): continue
        try:
            record = parse_student(line)
            valid_records.append(record)
        except ValueError as e:
            invalid_count += 1
            print(f"第{lineno}行无效: {line.strip()} -> 错误原因: {e}")

# 5. 写入报告文件
print("\n--- 5. 写入报告 ---")
avg = sum(r["score"] for r in valid_records) / len(valid_records)
rate = sum(1 for r in valid_records if r["score"] >= 60) / len(valid_records)
report_path = os.path.join(data_dir, "report.txt")
with open(report_path, "w", encoding="utf-8") as f:
    f.write(f"有效记录数: {len(valid_records)}, 无效记录数: {invalid_count}\n")
    f.write(f"平均分: {avg:.2f}, 及格率: {rate:.0%}\n")
# 演示 a 模式追加
with open(report_path, "a", encoding="utf-8") as f:
    f.write("追加一行演示 (a模式)\n")
print("报告已保存至:", report_path)

# 6. pickle 序列化
print("\n--- 6. pickle 保存与读取 ---")
pkl_path = os.path.join(data_dir, "students.pkl")
with open(pkl_path, "wb") as f:
    pickle.dump(valid_records, f)
with open(pkl_path, "rb") as f:
    loaded_records = pickle.load(f)
print("恢复的数据与原数据是否相等:", loaded_records == valid_records)

# 8. 读取不存在的文件捕获异常
print("\n--- 8. 捕获 FileNotFoundError ---")
try:
    with open("no_such_file.txt", encoding="utf-8") as f:
        pass
except FileNotFoundError as e:
    print(f"成功捕获异常: {e}")
```

```python
# ========= 实验5：类、继承与对象复制 =========
import copy

# 1. 定义 Student 类
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score

    def is_passed(self):
        return self.score >= 60

    def __str__(self):
        return f"Student({self.name}, {self.score})"

# 2. 为有效记录创建对象并输出
print("--- 2. 创建 Student 对象 ---")
students_obj = [Student(r["name"], r["score"]) for r in valid_records]
for s in students_obj:
    print(f"{s} -> 及格状态: {s.is_passed()}")

# 3. 定义 GraduateStudent 子类
class GraduateStudent(Student):
    def __init__(self, name, score, research_topic):
        super().__init__(name, score)
        self.research_topic = research_topic

    def __str__(self):
        return f"GradStudent({self.name}, {self.score}, 研究方向: {self.research_topic})"

print("\n--- 3. 测试继承 ---")
g_student = GraduateStudent("Alice", 85, "人工智能")
print(g_student)
print(f"是否及格: {g_student.is_passed()}")

# 4. 直接赋值、浅拷贝、深拷贝的区别
print("\n--- 4. 复制机制对比 ---")
# 注意每次比较都要从新建的原始对象开始
def make_orig():
    return {"name": "Alice", "scores": [85, 90]}

o1 = make_orig(); a1 = o1
o2 = make_orig(); c1 = copy.copy(o2)
o3 = make_orig(); d1 = copy.deepcopy(o3)

a1["scores"].append(100)
c1["scores"].append(100)
d1["scores"].append(100)

print(f"直接赋值 (o1 被改变): {o1}")
print(f"浅拷贝   (o2 内层列表被改变): {o2}")
print(f"深拷贝   (o3 保持不变): {o3}")
```
