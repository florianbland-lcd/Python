```python
# ========= 第一题：定义简单函数 =========
def say_hi():
    print("Hello, world!")

def goat_pad(s):
    print("goat" + s + "goat")

def print_n(s, n):
    for _ in range(n):
        print(s)

# 运行测试
print("--- 测试 say_hi ---")
say_hi()
print("\n--- 测试 goat_pad ---")
goat_pad("abc")
print("\n--- 测试 print_n ---")
print_n("你好", 3)
```

```python
# ========= 第二题：欧几里得算法 =========
def gcd(m, n):
    while n != 0:
        m, n = n, m % n
    return m

# 评估数字对
print("\n--- 计算最大公约数(GCD) ---")
print(f"gcd(2018, 2019) = {gcd(2018, 2019)}")
print(f"gcd(1200, 300)  = {gcd(1200, 300)}")
print(f"gcd(5040, 60)   = {gcd(5040, 60)}")

# 负数测试
print(f"gcd(-12, 8)     = {gcd(-12, 8)}")
```

```python
# ========= 第三题：近似欧拉数 e =========
import math

# 1. 极限公式
def euler_limit(n):
    return (1 + 1/n) ** n

# 2. 无限级数求和
def euler_infinite_sum(n):
    total = 0.0
    fact = 1
    for k in range(n):
        if k > 0:
            fact *= k
        total += 1.0 / fact
    return total

# 3. 误差逼近
def euler_approx(eps):
    true_e = math.exp(1)
    total = 0.0
    fact = 1
    k = 0
    while abs(total - true_e) > eps:
        if k > 0:
            fact *= k
        total += 1.0 / fact
        k += 1
    return total

# 4. 打印表格
def print_euler_sum_table(n):
    for k in range(1, n + 1):
        print(euler_infinite_sum(k))


# ===== 运行测试 =====
print("--- 测试 euler_limit(1000) ---")
print(euler_limit(1000))

print("\n--- 测试 euler_infinite_sum ---")
print(f"n=0: {euler_infinite_sum(0)}")  # 应为 0
print(f"n=1: {euler_infinite_sum(1)}")  # 应为 1
print(f"n=4: {euler_infinite_sum(4)}")  # 应约为 2.667

print("\n--- 测试 euler_approx(0.001) ---")
print(euler_approx(0.001))

print("\n--- 测试 print_euler_sum_table(5) ---")
print_euler_sum_table(5)
```

```python
# ========= 第四题：测试整数属性 =========

# 1. 判断是否是 2 的幂
def is_power_of_2(n):
    if n < 1:
        return False
    while n % 2 == 0:
        n //= 2
    return n == 1

# 2. 判断是否是 b 的幂
def is_power(b, n):
    if n < 1:
        return False
    while n % b == 0:
        n //= b
    return n == 1

# ===== 运行测试 =====
print("--- 测试 is_power_of_2 ---")
print(f"is_power_of_2(1)  = {is_power_of_2(1)}")   # True (2的0次方)
print(f"is_power_of_2(4)  = {is_power_of_2(4)}")   # True
print(f"is_power_of_2(6)  = {is_power_of_2(6)}")   # False
print(f"is_power_of_2(16) = {is_power_of_2(16)}")  # True

print("\n--- 测试 is_power(b, n) ---")
print(f"is_power(3, 9)   = {is_power(3, 9)}")      # True (3的2次方)
print(f"is_power(3, 27)  = {is_power(3, 27)}")     # True
print(f"is_power(3, 10)  = {is_power(3, 10)}")     # False
print(f"is_power(5, 25)  = {is_power(5, 25)}")     # True
```
